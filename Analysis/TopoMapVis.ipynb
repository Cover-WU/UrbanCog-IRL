```python
shenzhen_grid = gpd.read_file('./data/ss_city_grid/ss_city_grid_by_cover.shp')
shenzhen_area = shenzhen_grid.dissolve()
```

```python
def trajVis(who, week_order, ax):
    od_pairs = trajFiner.traj4Visualize(who, week_order)
    od_counter = Counter(od_pairs)
    # 创建地理坐标系的地图
    shenzhen_area.plot(ax=ax, color='lightgray', edgecolor='black')
    # 绘制弯曲的箭头
    for flow, count in od_counter.items():
        start, end = flow
        lon_start, lat_start = start
        lon_end, lat_end = end
        
        # 定义弯曲路径的控制点
        control_lon = (lon_start + lon_end) / 2  # 经度中点
        control_lat = (lat_start + lat_end) / 2 + (lat_end - lat_start) * 0.5  # 纬度中点上移，增加弯曲度

        # 创建贝塞尔曲线路径
        verts = [
            (lon_start, lat_start),  # 起点
            (control_lon, control_lat),  # 控制点
            (lon_end, lat_end)  # 终点
        ]
        codes = [Path.MOVETO, Path.CURVE3, Path.CURVE3]  # 使用二次贝塞尔曲线
        path = Path(verts, codes)

        # 创建并绘制路径上的箭头
        patch = FancyArrowPatch(
            path=path,
            color='blue',
            linewidth=1 + (count-1)/2,
            arrowstyle='->',  # 箭头样式
            mutation_scale=10  # 控制箭头大小
        )
        ax.add_patch(patch)
```

```python
def cogGraphVis(who, week_order, ax):
    G = trajFiner.cogGraph4Visualize(who, week_order)    
    # 创建地理坐标系的地图
    shenzhen_area.plot(ax=ax, color='lightgray', edgecolor='black')
    
    # Extract and scale weights for the edges
    weights = [G[u][v]['weight'] for u, v in G.edges()]
    print("weights: ", weights)
    alphaw = np.interp(weights, (min(weights), max(weights)), (0, 1))
    # Extract the size weighs for the nodes
    node_size = [G.nodes[n]['weight'] for n in G.nodes()]
    print("node_size: ", node_size)
    node_size = np.log(np.array(node_size))
    node_size = np.interp(node_size, (min(node_size), max(node_size)), (10, 100))
    # Position
    pos = nx.get_node_attributes(G, 'pos')
    # Plot the network on the map
    nx.draw_networkx_nodes(G, pos, ax=ax, node_size=node_size, node_color='orange')
    # 绘制地图上的网络边，使用弯曲的边防止遮挡
    for u, v, data in G.edges(data=True):
        weight = data['weight']
        alpha = np.interp(weight, (min(weights), max(weights)), (0, 1))
        
        # 获取起点和终点的位置
        start = pos[u]
        end = pos[v]
        
        # 定义弯曲路径的控制点
        control_lon = (start[0] + end[0]) / 2  # 经度中点
        control_lat = (start[1] + end[1]) / 2 + (end[1] - start[1]) * 0.2  # 纬度中点上移，增加弯曲度

        # 创建贝塞尔曲线路径
        verts = [
            (start[0], start[1]),  # 起点
            (control_lon, control_lat),  # 控制点
            (end[0], end[1])  # 终点
        ]
        codes = [Path.MOVETO, Path.CURVE3, Path.CURVE3]  # 使用二次贝塞尔曲线
        path = Path(verts, codes)

        # 创建并绘制路径上的箭头
        patch = FancyArrowPatch(
            path=path,
            color=plt.cm.Reds(weight / max(weights)),
            linewidth=2.5,
            alpha=alpha,
            arrowstyle='-|>',  # 箭头样式
            mutation_scale=10  # 控制箭头大小
        )
        ax.add_patch(patch)
```

```python
def cogGraphEvolution(who):
    visit_dates = SIRLU.visited_date(who)
    _, week_code = SIRLU.extract_week_ends(visit_dates)

    fig_reservoir = list()
    week_code_unique = np.unique(week_code)
    for week_order in week_code_unique:
        fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(20, 8))
        trajVis(who, week_order, ax1)
        cogGraphVis(who, week_order, ax2)
        fig.suptitle(f"Trajectory and CoG Graph on week {week_order}", fontsize=16)
        fig_reservoir.append(fig)
    return fig_reservoir
```

```python
def topoSpectVis(who):
    spectral_label = trajFiner.topoNodeCluster(who, method='hdbscan', optimal='silhouette')
    id_coords_mapping = SIRLU.load_id_coords_mapping(who)
    id_coords_mapping_sorted = dict(sorted(id_coords_mapping.items()))
    id_month = SIRLU.load_id_month_mapping(who)
    id_geometry = {k: Point(*v) for k, v in id_coords_mapping_sorted.items()}
    gpd_id_geometry = gpd.GeoDataFrame.from_dict(id_geometry, orient='index', columns=['geometry'], crs='EPSG:4326')
    gpd_id_geometry['idx'] = gpd_id_geometry.index
    gpd_id_geometry['label'] = spectral_label
    gpd_id_geometry['month'] = gpd_id_geometry.index.map(id_month)
    # explore the results
    m = gpd_id_geometry.explore(column='label', cmap='tab20', marker_kwds={'radius': 8},)
    return m 
```

```python
import dash
from dash import dcc, html
from dash.dependencies import Input, Output
import plotly.express as px


# 创建 Dash 应用
app = dash.Dash(__name__)

# 获取所有可用的用户ID
def get_available_users():
    # 这里需要根据你的实际情况获取用户列表
    # 例如从某个目录或数据库中获取
    user_list = [68058890, 10013454, 58272403,
                  6945721, 71209087, 93854949,
                 82455786, 58124481, 54636959,
                  1102234, 25679537,102181433]  # 示例用户列表
    user_list.sort()
    return user_list

# 定义应用布局
app.layout = html.Div([
    # change color to white
    html.H1("Visualization of Simple Clustering", style={'color': 'white'}),
    
    # 添加用户选择下拉框
    html.Div([
        html.Label("Choose user ID: ", style={'color': 'white'}),
        dcc.Dropdown(
            id='user-dropdown',
            options=[{'label': f'User {user}', 'value': user} 
                    for user in get_available_users()],
            value=get_available_users()[0]  # 默认选择第一个用户
        )
    ], style={'width': '30%', 'margin': '10px'}),
    
    # 添加聚类方法选择
    html.Div([
        html.Label("Choose clustering method: ", style={'color': 'white'}),
        dcc.Dropdown(
            id='cluster-method-dropdown',
            options=[
                {'label': 'Knee', 'value': 'knee'},
                {'label': 'HDBSCAN', 'value': 'hdbscan'}
            ],
            value='knee'  # 默认选择knee方法
        )
    ], style={'width': '30%', 'margin': '10px'}),
    
    # 地图显示区域
    html.Div(id='map-container')
])

# 回调函数：根据选择的用户ID更新地图
@app.callback(
    Output('map-container', 'children'),
    [Input('user-dropdown', 'value'),
     Input('cluster-method-dropdown', 'value')]
)
def update_map(selected_user, selected_method):
    if not selected_user:
        return html.Div("Please choose a user ID")
    
    try:
        # 执行聚类
        cluster_labels = topoMap.clusterLocationsSimple(selected_user, clusterer=selected_method)
        
        # 准备数据
        id_coords_mapping = SIRLU.load_id_coords_mapping(selected_user)
        id_coords_mapping_sorted = dict(sorted(id_coords_mapping.items()))
        id_month = SIRLU.load_id_month_mapping(selected_user)
        
        # 创建GeoDataFrame
        points_data = []
        for k, v in id_coords_mapping_sorted.items():
            points_data.append({
                'geometry': Point(*v),
                'id': k,
                'label': str(cluster_labels[k]),
                'month': id_month[k],
                'lat': v[1],
                'lon': v[0]
            })
        
        gdf = gpd.GeoDataFrame(points_data)
        
        # 使用Plotly创建地图
        fig = px.scatter_mapbox(
            gdf,
            lat='lat',
            lon='lon',
            color='label',
            hover_data=['id', 'month'],
            zoom=11,
            color_discrete_sequence=px.colors.qualitative.Light24
        )
        
        # 设置地图样式
        fig.update_layout(
            mapbox_style="carto-positron",
            margin={"r":0,"t":0,"l":0,"b":0},
            height=500
        )
        
        return dcc.Graph(figure=fig)
    
    except Exception as e:
        return html.Div(f"Error: {str(e)}", style={'color': 'white'})

app.run(debug=True, port=8050)
```

## Check the performance of the day-level clustering method

```python
# 导入必要的库
import geopandas as gpd
from shapely.geometry import Point
import pickle
import os
from datetime import datetime
import topoMap
import glob
import json
import numpy as np
from matplotlib.colors import LinearSegmentedColormap

def get_id_month_mapping(who):
    """
    获取每个id对应的月份
    """
    # 读取轨迹数据
    json_path = f'data/user_data_survey/{who:09d}/all_traj.json'
    with open(json_path, 'r') as f:
        traj_data = json.load(f)
    
    # 创建id到月份的映射
    id_month = {}
    for tc in traj_data:
        ids = tc['id_chain']
        date = str(tc['date'])
        month = date[:6]
        for id in ids:
            if id not in id_month:
                id_month[int(id)] = month
    return id_month

# 修改 create_distinct_colormap 函数
def create_distinct_colormap(n_clusters):
    """
    创建一个视觉上容易区分的颜色映射，使用matplotlib的tab20调色板
    """
    from matplotlib import cm
    from matplotlib.colors import to_hex

    # 使用tab20调色板生成颜色
    cmap = cm.get_cmap('tab20', n_clusters)
    hex_colors = [to_hex(cmap(i)) for i in range(n_clusters)]
    return hex_colors

def visualize_weekly_clusters(who):
    """
    为每周的聚类结果创建交互式地图
    """
    # 获取id到月份的映射
    id_month = get_id_month_mapping(who)
    
    # 获取所有已处理的周数据文件
    res_dir = topoMap.topoResPath(who)
    pickle_files = glob.glob(os.path.join(res_dir, 'topo_res_*.pickle'))
    
    # 按日期排序
    pickle_files.sort()
    
    # 为每个文件创建地图
    for file_path in pickle_files:
        # 从文件名提取日期
        date_str = os.path.basename(file_path).replace('topo_res_', '').replace('.pickle', '')
        end_date = int(date_str)
        print(f"\n处理周结束日期: {datetime.strptime(str(end_date), '%Y%m%d').strftime('%Y-%m-%d')}")
        
        # 读取聚类结果
        with open(file_path, 'rb') as f:
            _, id_coorders_mapping_edit, _, cluster_labels = pickle.load(f)
        
        # 创建GeoDataFrame，使用pickle文件中的id_coords_mapping
        id_geometry = {k: Point(*v) for k, v in id_coorders_mapping_edit.items()}
        gdf = gpd.GeoDataFrame.from_dict(id_geometry, orient='index', 
                                       columns=['geometry'], crs='EPSG:4326')
        gdf['cluster'] = cluster_labels
        
        # 添加月份信息
        gdf['month'] = gdf.index.map(id_month)
        gdf['idx'] = gdf.index
        
        # 创建自定义颜色映射
        n_clusters = len(set(cluster_labels))
        cmap = create_distinct_colormap(n_clusters)
        
        # 创建一个颜色映射字典
        cluster_to_color = {cluster: cmap[i] for i, cluster in enumerate(set(cluster_labels))}
        
        # 使用explore创建交互式地图
        m = gdf.explore(
            column='cluster',
            tooltip=['cluster', 'month', 'idx'],  # 添加月份到提示信息
            tiles='OpenStreetMap',
            cmap='tab20',
            style_kwds={
                'radius': 8,
                'fillOpacity': 0.6,
            },
            popup=True,
            legend=True,
            legend_kwds={
                'title': '聚类编号',
                'loc': 'upper right'
            },
            title=f'用户{who}在{datetime.strptime(str(end_date), "%Y%m%d").strftime("%Y-%m-%d")}的聚类结果'
        )
        
        print(f"该周共有 {len(cluster_labels)} 个位置点，{n_clusters} 个聚类")
        display(m)
            
# 运行可视化
who = 1102234
visualize_weekly_clusters(who)
```
